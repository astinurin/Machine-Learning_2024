```python
import matplotlib.pyplot as plt

gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

gray_image.shape

face_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
face = face_classifier.detectMultiScale(gray_image, scaleFactor=1.1, minNeighbors=5, minSize=(40, 40))

for (x, y, w, h) in face:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 4)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

import matplotlib.pyplot as plt

plt.figure(figsize=(20,10))
plt.imshow(img_rgb)
plt.axis('off')
```

```python

# Path gambar
imagePath = '/content/drive/My Drive/Machine_Learning/tryFaceDetection/sb-a.jpg'

# Membaca gambar
img = cv2.imread(imagePath)

# Memeriksa apakah gambar berhasil dibaca
if img is None:
    print("Gambar tidak ditemukan. Periksa path gambar.")
else:
    # Mengonversi gambar ke grayscale
    gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Memuat classifier Haar Cascade untuk deteksi mata
    eye_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_eye.xml')

    # Deteksi mata pada gambar
    eyes = eye_classifier.detectMultiScale(gray_image)

    # Gambar kotak di sekitar mata yang terdeteksi
    for (ex, ey, ew, eh) in eyes:
        cv2.rectangle(img, (ex, ey), (ex + ew, ey + eh), (255, 0, 0), 4)

    # Mengonversi gambar dari BGR ke RGB untuk menampilkan dengan matplotlib
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Menampilkan gambar dengan deteksi mata
    plt.figure(figsize=(20, 10))
    plt.imshow(img_rgb)
    plt.axis('off')  # Menghilangkan axis
    plt.show()
```
